```python
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
```

```python
from langchain_core.documents import Document

# Create Langchain documents for IPL players

doc1 = Document(page_content="Rohit Sharma is a cricketer", metadata={"source": "Rohit Sharma"})
doc2 = Document(page_content="Virat Kohli is a good cricketer", metadata={"source": "Virat Kohli"})
doc3 = Document(page_content="MS Dhoni captain", metadata={"source": "MS Dhoni"})
doc4 = Document(page_content="Sachin Tendulkar is a cricketer", metadata={"source": "Sachin Tendulkar"})
doc5 = Document(page_content="Sachin Tendulkar is a cricketer", metadata={"source": "Sachin Tendulkar"})
```

```python
docs = [doc1, doc2, doc3 ,doc4, doc5]
```

```python
import os
# Set your OpenAI API key as an environment variable
# Replace 'YOUR_OPENAI_API_KEY' with your actual API key
os.environ["OPENAI_API_KEY"] = "YOUR_OPENAI_API_KEY"

vector_store = Chroma(
    embedding_function=OpenAIEmbeddings(),
    persist_directory="./chroma_db",
    collection_name="my_collection"
)
```

```python
# add documents
vector_store.add_documents(docs)
```

```python
# view documents
vector_store.get(include= ['embeddings', 'documents', 'metadates'])
```

```python
# search documents
vector_store.similarity_search(query='who among these are a bowler?',
                                k=2)

```

```python
# search with similarity score
vector_store.similarity_search_with_score(query='who among these are a bowler?',
                                            k=2)
```

```python
# meta_data filtering
vector_store.similarity_search_with_score(query='',

                                filter={'team': 'Chennai Super Kings'})
```

```python
#update documents
updated_doc1 = Document(page_content="Rohit Sharma is a cricketer", metadata={"source": "Rohit Sharma"})

vector_store.update_document(document_id='fdb1', document = updated_doc1)
```

```python
# view documents
vector_store.get(include= ['embeddings', 'documents', 'metadates'])
```

```python
# delete  document
vector_store.delete_document(ids=['fdb1'])
```
